```python
#Ques24- Calculate electricity bill using conditions.
'''0–50 Units: Units × ₹3.21
51–100 Units: (50 × ₹3.21) + (Remaining units × ₹4.67)
101–200 Units: (50 × ₹3.21) + (50 × ₹4.67) + (Remaining units × ₹6.66)
>200 Units: (50 × ₹3.21) + (50 × ₹4.67) + (100 × ₹6.66) + (Remaining units × ₹8.50)'''
units=int(input("Enter the number of units consumed: "))
if units<=50:
  bill=units*3.21
elif units<=100:
  bill=(50*3.21)+(units-50)*4.6
elif units<=200:
  bill=(50*3.21)+(50*4.6)+((units-100)*6.6)
else:
  bill=(50*3.21)+(50*4.6)+(100*6.6)+((units-200)*8.5)
print(f"Electricity bill = ₹{bill:.2f}")

```

```python
#Ques53- Print number pyramid pattern.
n = int(input("Enter the number of rows: "))
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    for j in range(i - 1, 0, -1):
        print(j, end=" ")
    print()
```

```python
#Ques54- Print hollow square pattern.
n=int(input("Enter the size of the square: "))
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n- 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
#Ques55- Print Pascal’s triangle.
rows = int(input("Enter number of rows: "))
for i in range(rows):
    for j in range(rows - i - 1):
        print(" ", end=" ")
    num = 1
    for j in range(i + 1):
        print(num, end="   ")
        num = num * (i - j) // (j + 1)
    print()
```

```python
#Ques91- Create a recursive function for factorial.
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
```

```python
#Ques92- Create a function to find Fibonacci series.
def fibonacci(n,a=0,b=1):
  if n<=0:
    return[]
  elif n==1:
    return[0]
  series=[0,1]

  while len(series) < n:
    next_term = series[-1] + series[-2]
    series.append(next_term)
  return series

print(fibonacci(10))
```

```python
#Ques99- Create a Student class with attributes and methods.
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks


    def display(self):
        print("Student Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)

s1 = Student("Shakshi", 22, 95)
s1.display()
```

```python
#Ques100- Build a simple ATM system using Python.
# Simple ATM System

balance = 10000

while True:
    print("\n===== ATM MENU =====")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("Current Balance = ₹", balance)

    elif choice == 2:
        deposit = float(input("Enter deposit amount: ₹"))
        balance += deposit
        print("Amount Deposited Successfully")
        print("Updated Balance = ₹", balance)

    elif choice == 3:
        withdraw = float(input("Enter withdrawal amount: ₹"))

        if withdraw <= balance:
            balance -= withdraw
            print("Withdrawal Successful")
            print("Remaining Balance = ₹", balance)
        else:
            print("Insufficient Balance")

    elif choice == 4:
        print("Thank you for using ATM")
        break

    else:
        print("Invalid Choice")
```
